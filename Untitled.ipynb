```python
import pandas as pd
import matplotlib.pyplot as plt


# Read the dataset
sales1 = pd.read_csv("sales1.csv")


# Display the dataset
print(sales1)


# Find number of rows and columns
print("Shape of dataset:")
print(sales1.shape)


# Display information about dataset
print("\nDataset Information:")
sales1.info()


# Check missing values
print("\nMissing Values:")
print(sales1.isnull().sum())


# Statistical information
print("\nStatistical Values:")
print(sales1.describe())


# Check duplicate rows
print("\nNumber of Duplicate Rows:")
print(sales1.duplicated().sum())


# Remove duplicate rows
sales1 = sales1.drop_duplicates()


# Check data types
print("\nData Types:")
print(sales1.dtypes)


# Convert Sale_Date into datetime format
sales1["Sale_Date"] = pd.to_datetime(
    sales1["Sale_Date"],
    errors="coerce"
)

sales1["Month"] = sales1["Sale_Date"].dt.month_name()

# -------------------------------
# Region-wise Sales
# -------------------------------

sales1.groupby("Region")["Sales"].sum().plot(
    kind="bar",
    ylabel="Total Sales",
    title="Region Wise Sales"
)

plt.show()


# -------------------------------
# Product-wise Sales
# -------------------------------

sales1.groupby("Product")["Sales"].sum().plot(
    kind="bar",
    ylabel="Total Sales",
    title="Product Wise Sales"
)

plt.show()


# -------------------------------
# Category-wise Sales
# -------------------------------

sales1.groupby("Category")["Sales"].sum().plot(
    kind="bar",
    ylabel="Total Sales",
    title="Category Wise Sales"
)

plt.show()


# -------------------------------
# Payment Method-wise Sales
# -------------------------------

sales1.groupby("Payment_Method")["Sales"].sum().plot(
    kind="pie",
    autopct="%1.1f%%",
    title="Payment Method Wise Sales"
)

plt.ylabel("")
plt.show()


# -------------------------------
# Create Month column
# -------------------------------

sales1["Month"] = sales1["Sale_Date"].dt.month_name()


# Display updated dataset
print(sales1)


# -------------------------------
# Month-wise Sales
# -------------------------------

sales1.groupby("Month")["Sales"].sum().plot(
    kind="bar",
    ylabel="Total Sales",
    title="Month Wise Sales"
)

plt.show()
```
